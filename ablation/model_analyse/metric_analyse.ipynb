```python
# reading cd hit

with open('../../phase_2_model.json') as f:
    epoch_data = json.load(f)

    best_val_func_mcc = 0
    best_epoch = None
    
    for e in epoch_data:
        if best_val_func_mcc <= e['val_avg_func_mcc']:
            best_val_func_mcc = e['val_avg_func_mcc']
            best_epoch = e

phase_2_data = best_epoch
print(best_val_func_mcc)
```

```python
name_mapping = {
    'NON-FUNCTIONAL': 'Non-Functional',
    'anti-bacterial': 'Anti-Bacterial',
    'anti-viral': 'Anti-Viral',
    'anti-fungal': 'Anti-Fungal',
    'anti-parasitic': 'Anti-Parasitic',
    'anti-cancer': 'Anti-Cancer',
    'toxic': 'Toxic',
    'metabolic': 'Metabolic',
    'signal-peptide': 'Signal-Peptide',
    'cell-cell-communication': 'Cell-Cell Comm.',
    'immunological': 'Immunological',
    'inhibitor': 'Inhibitor',
    'drug-delivery': 'Drug-Delivery',
    'other-functional': 'Other-Functional'
}
# Desired order to match your example table
order = [
    'NON-FUNCTIONAL', 'anti-bacterial', 'anti-viral', 'anti-fungal', 
    'anti-parasitic', 'anti-cancer', 'toxic', 'metabolic', 
    'signal-peptide', 'cell-cell-communication', 'immunological', 
    'inhibitor', 'drug-delivery', 'other-functional'
]

# Construct the LaTeX strings
latex_lines = [
    "\\begin{table}[hbt!]",
    "\\centering",
    "\\caption{Detailed Performance Metrics by Functional Category on the Independent Test Set across Training Phases}",
    "\\label{tab:performance_metrics}",
    "\\setlength{\\tabcolsep}{5pt}"
    "\\begin{tabular}{lccccccc}",
    "\\toprule",
    "\\multirow{2}{*}{\\textbf{Functional Category}}  & \\multicolumn{6}{c}{\\textbf{Phase 2 (Length-Corrected Evaluation)}} & \\textbf{Phase 1} \\\\"
    "\\cmidrule(lr){2-7} \\cmidrule(lr){8-8}& \\textbf{Accuracy} & \\textbf{Precision} & \\textbf{Recall} & \\textbf{F1-Score} & \\textbf{ROC AUC} & \\textbf{MCC} & \\textbf{MCC} \\\\"
    "\\midrule"
]
all_acc = []
all_prec = []
all_rec = []
all_f1 = []
all_roc = []
all_mcc = []
all_phase_1_mcc = []
for key in order:
    if key in phase_2_data['test_metrics']:
        metrics = phase_2_data['test_metrics'][key]
        phase_1_metric = phase_1_data['test_metrics'][key]
        name = name_mapping.get(key, key)
        
        # Format metrics to 3 decimal places
        acc = f"{metrics['accuracy']:.3f}"
        all_acc.append(acc)
        prec = f"{metrics['precision']:.3f}"
        all_prec.append(prec)
        rec = f"{metrics['recall']:.3f}"
        all_rec.append(rec)
        f1 = f"{metrics['f1_score']:.3f}"
        all_f1.append(f1)
        mcc = f"{metrics['mcc']:.3f}"
        all_mcc.append(mcc)
        mcc_phase_1 = f"{phase_1_metric['mcc']:.3f}"
        all_phase_1_mcc.append(mcc_phase_1)
        roc = f"{metrics['roc_auc']:.3f}"
        all_roc.append(roc)
        
        
        row_str = f"{name} & {acc} & {prec} & {rec} & {f1} & {roc} & {mcc} & {mcc_phase_1}\\\\"
        latex_lines.append(row_str)


all_acc = np.array(all_acc, dtype=float)
all_prec = np.array(all_prec, dtype=float)
all_rec = np.array(all_rec, dtype=float)
all_f1 = np.array(all_f1, dtype=float)
all_roc = np.array(all_roc, dtype=float)
all_mcc = np.array(all_mcc, dtype=float)
all_phase_1_mcc = np.array(all_phase_1_mcc, dtype=float)

latex_lines.extend([
    "\\bottomrule",
    f"Mean & {all_acc.mean():.3f} & {all_prec.mean():.3f} & {all_rec.mean():.3f} & {all_f1.mean():.3f} & {all_roc.mean():.3f} & {all_mcc.mean():.3f} & {all_phase_1_mcc.mean():.3f}\\\\",
    "\\end{tabular}",
    "\\end{table}"
])

# Print out the final LaTeX code
print("\n".join(latex_lines))
```

```python
def dict_mean(d):
    return sum([v for v in d.values()])/len(d)
```

```python
name_mapping = {
    'NON-FUNCTIONAL': 'Non-Functional',
    'anti-bacterial': 'Anti-Bacterial',
    'anti-viral': 'Anti-Viral',
    'anti-fungal': 'Anti-Fungal',
    'anti-parasitic': 'Anti-Parasitic',
    'anti-cancer': 'Anti-Cancer',
    'toxic': 'Toxic',
    'metabolic': 'Metabolic',
    'signal-peptide': 'Signal-Peptide',
    'cell-cell-communication': 'Cell-Cell Comm.',
    'immunological': 'Immunological',
    'inhibitor': 'Inhibitor',
    'drug-delivery': 'Drug-Delivery',
    'other-functional': 'Other-Functional'
}
# Desired order to match your example table
order = [
    'NON-FUNCTIONAL', 'anti-bacterial', 'anti-viral', 'anti-fungal', 
    'anti-parasitic', 'anti-cancer', 'toxic', 'metabolic', 
    'signal-peptide', 'cell-cell-communication', 'immunological', 
    'inhibitor', 'drug-delivery', 'other-functional'
]
print("Classes    0.8_cd_hit 0.9_cd_hit no_cd_hit cnn_removed esm_removed")
for k in esm_removed_mcc.keys():
    print(k, f"{phase_1_mcc[k]:.3f}", f"{cd_hit_9_mcc[k]:.3f}", f"{cd_hit_no_mcc[k]:.3f}", f"{cnn_removed_mcc[k]:.3f}", f"{esm_removed_mcc[k]:.3f}")

print("Mean", f"{dict_mean(phase_1_mcc):.3f}", f"{dict_mean(cd_hit_9_mcc):.3f}", f"{dict_mean(cd_hit_no_mcc):.3f}", f"{dict_mean(cnn_removed_mcc):.3f}", f"{dict_mean(esm_removed_mcc):.3f}")
```

```python
# --- LaTeX table ---
models = [
    ("CD-HIT(Threshold 0.8)", phase_1_mcc),
    ("CD-HIT(Threshold 0.9)", cd_hit_9_mcc),
    ("Without CD-HIT", cd_hit_no_mcc),
    ("CNN Removed", cnn_removed_mcc),
    ("ESM Removed", esm_removed_mcc),
]

col_spec = "l" + "c" * len(models)
print(r"\begin{table}[hbt!]")
print(r"\centering")
print(r"\caption{Ablation study (Phase-1 curriculum). Per-class MCC under varying CD-HIT threshold and multi-representation removal. The baseline configuration (CD-HIT(Threshold 0.8), full model) is highlighted.}")
print(r"\label{tab:ablation_mcc}")
print(r"\setlength{\tabcolsep}{5pt}")
print(r"\begin{tabular}{lccccc}")
print(r"\toprule")
print(r"Class & CD-HIT(0.8) & CD-HIT(0.9) & No CD-HIT & CNN Removed & ESM Removed \\")
print(r"\midrule")

for cls in order:
    display = name_mapping.get(cls, cls)
    vals = " & ".join(f"{d[cls]:.3f}" for _, d in models)
    print(f"{display} & {vals} \\\\")

print(r"\midrule")
mean_vals = " & ".join(f"{dict_mean(d):.3f}" for _, d in models)
print(f"Mean & {mean_vals} \\\\")
print(r"\bottomrule")
print(r"\end{tabular}")
print(r"\end{table}")
```

```python
# --- LaTeX table: phase_1 vs mcmfpp vs slfe vs cfec ---
# phase_1_mcc uses 'NON-FUNCTIONAL' (uppercase), others use 'non-functional' (lowercase)
def get_mcc(d, cls):
    """Look up MCC, trying both the original key and lowercased variant."""
    if cls in d:
        return d[cls]
    # fallback: try lowercase (for phase_1_mcc -> NON-FUNCTIONAL vs non-functional)
    lower = cls.lower()
    if lower in d:
        return d[lower]
    return float('nan')

models2 = [
    ("PepMTL (Phase 1)", phase_1_mcc),
    ("MCMFPP", mcmfpp),
    ("SLFE", slfe),
    ("CFEC", cfec),
]

col_spec2 = "l" + "c" * len(models2)
print(r"\begin{table}[hbt!]")
print(r"\centering")
print(r"\caption{Head-to-head comparison of PepMTL with MCMFPP on our homology-aware (CD-HIT 0.8) test split}")
print(r"\label{tab:mcmfpp_benchmark}")
print(r"\setlength{\tabcolsep}{5pt}")
print(r"\begin{tabular}{" + col_spec2 + r"}")
print(r"\toprule")
print("Class & " + " & ".join(name for name, _ in models2) + r" \\")
print(r"\midrule")

for cls in order:
    display = name_mapping.get(cls, cls)
    vals = " & ".join(f"{get_mcc(d, cls):.3f}" for _, d in models2)
    print(f"{display} & {vals} \\\\")

print(r"\midrule")
mean_vals = " & ".join(f"{dict_mean(d):.3f}" for _, d in models2)
print(f"Mean & {mean_vals} \\\\")
print(r"\bottomrule")
print(r"\end{tabular}")
print(r"\end{table}")
```
